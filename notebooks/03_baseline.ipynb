```python
# 1. Завантаження готових фічей

import pickle

with open("/content/drive/MyDrive/project_data/features_engineered.pkl", "rb") as f:
    data = pickle.load(f)

X_train = data["X_train"]
X_val = data["X_val"]
X_test = data["X_test"]
y_train = data["y_train"]
y_val = data["y_val"]
y_test = data["y_test"]

print(X_train.shape, X_val.shape, X_test.shape)

# 2. Baseline модель (найчастіший клас)
from sklearn.dummy import DummyClassifier
from sklearn.metrics import classification_report, f1_score, accuracy_score

baseline = DummyClassifier(strategy="most_frequent")
baseline.fit(X_train, y_train)

y_pred_val = baseline.predict(X_val)

print("=== Baseline (найчастіший клас) — Validation ===")
print(classification_report(y_val, y_pred_val))
print("Macro F1:", f1_score(y_val, y_pred_val, average="macro"))
print("Accuracy:", accuracy_score(y_val, y_pred_val))

# 3. Зберігаємо результати baseline для порівняльної таблиці експериментів
results = {
    "model": "Baseline (most frequent)",
    "params": "-",
    "val_accuracy": accuracy_score(y_val, y_pred_val),
    "val_macro_f1": f1_score(y_val, y_pred_val, average="macro"),
}

import pandas as pd
results_df = pd.DataFrame([results])
results_df.to_csv("/content/drive/MyDrive/project_data/experiments_table.csv", index=False)
print(results_df)
```
